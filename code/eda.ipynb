### #3: Data preparation

```python
df.head()
```

```python
# Drop unnecessary column
df = df[[#'Unnamed: 0', 
    'Company', 'TypeName', 'Inches', 'ScreenResolution',
    'Cpu', 'Ram', 'Memory', 'Gpu', 'OpSys', 'Weight', 'Price']].copy()
```

```python
df = df.rename(columns={'TypeName':'Type_Name',
                        'ScreenResolution':'Screen_Resolution',
                        'Cpu':'CPU',
                        'Ram':'RAM',
                        'Gpu':'GPU',
                        'OpSys':'Operating_System'})
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

### #4 Feature Engineering

```python
# Extract CPU brand name
df['CPU_brand'] = df['CPU'].str.split().str[0]
```

```python
# Extract raw RAM number
df['RAM_gb'] = df['RAM'].str.replace('GB','').astype(int)
```

```python
# Extract GPU brand name
df['GPU_brand'] = df['GPU'].str.split().str[0]
```

```python
# Extract GPU specific model
df['GPU_model'] = df['GPU'].str.split(n=1).str[1]
```

```python
# Extract laptops' weight
df['Weight_kg'] = df['Weight'].str.replace('kg','').astype(float)
```

### #5: EDA

```python
sns.heatmap(df[['Inches', 'RAM_gb', 'Weight_kg', 'Price']].corr(), annot=True)
```

```python
for col in ['Company','Type_Name','CPU_brand','GPU_brand',]:
    sns.boxplot(x=col, y='Price',data=df); plt.xticks(rotation=45); plt.show()
```

```python
sns.boxplot(x='RAM_gb', y='Price', data=df)
```

```python
sns.regplot(x='Weight_kg', y='Price', data=df, scatter_kws={'alpha':0.3})
```

```python
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.histplot(df['Price'], bins=50, kde=True)
plt.title('Price Distribution')

plt.subplot(1, 2, 2)
sns.boxplot(x=df['Price'])
plt.title('Price Boxplot (Outliers)')
plt.tight_layout()
plt.show()
```

```python
Q1, Q3 = df['Price'].quantile([0.25, 0.75])
IQR = Q3 - Q1
print(f"Outliers below: {Q1 - 1.5*IQR:.0f}, above: {Q3 + 1.5*IQR:.0f}")
print(f"Outlier count: {((df['Price'] < Q1 - 1.5*IQR) | (df['Price'] > Q3 + 1.5*IQR)).sum()}")
```
